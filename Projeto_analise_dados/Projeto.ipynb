```python
# 1 - importação de bibliotecas:
import pandas as pd # Módulo para a leitura da base de dados
#import plotly.express as px # Módulo para a representaçao gráfica da base de dados
```

```python
# 2 - ler a base de dados:
tabela = pd.read_csv('retail_sales_dataset.csv')
```

```python
# 2.1 - imprimir a base de dados:
display(tabela)
```

```python
# 2.2 - Imprimir as informações da tabela
tabela.info()
```

Observações do tratamento:

1) A tabela possui duas informações irrelevantes para a análise exploratória de dados: Transaction ID e Customer ID.
2) A tabela possui uma variável com um tipo de dado que excede a sua necessidade de espaço na memória: Age.

```python
# 3 - Tratar a base de dados
for coluna in tabela.columns: # Adicionando _ no local de espaço em branco e modificando as letras das colunas para minúsculo.

    coluna_nova = coluna.split()
    coluna_nova = '_'.join(coluna_nova)

    coluna_nova = coluna_nova.lower()

    tabela = tabela.rename(columns={coluna: coluna_nova})

tabela = tabela.drop(columns=['transaction_id', 'customer_id']) # Descartando as colunas desnecessárias

tabela['age'] = tabela['age'].astype('int8') # Alterando o tipo de dado int64 para int8

# 3.1 - imprimir a tabela depois de modificada:
display(tabela)
tabela.info()
```

Conclusões de tratamento

1) Redução do tempo de percurso pela tabela
2) Redução da ocupação de memória da tabela em aproximadamente 31,82% (70.4 -> 48.0), que pode ser um valor alto para tabelas muito grandes


```python
# 4 - Verificar a relevância dos dados
tabela['product_category'].value_counts()
```

```python
tabela['product_category'].value_counts(normalize=True)
```

Ao examinar essa distribuição, percebe-se que há a tendência da empresa vender mais produtos da categoria "Clothing" e "Electronics" no período de exercício, porém a categoria "Beauty" não é irrelevante a ponto de se planejar previsões ou mobilizações substanciais.

```python
tabela['age'].value_counts() # Descobrindo a distribuição de idade
```

```python
# Análise: Percentual da faixa etária
percentual_idosos = (tabela['age'] >= 50).mean() * 100 # (tabela['age'] >= 50) -> acumulador de True, que é ni .mean() -> ni / n (distribuição relativa)
percentual_jovens = ((tabela['age'] >= 15) & (tabela['age'] <= 24)).mean() * 100
percentual_jovens_adultos = ((tabela['age'] > 24) & (tabela['age'] < 50)).mean() * 100
print('percentual de idade dos clientes: \n') 
print(percentual_idosos)
print(percentual_jovens_adultos)
print(percentual_jovens)
```

Conclusão: Nota-se que mais da metade dos clientes estão entre 24 e 50 anos de idade 

```python
# Distribuição do sexo biológico
tabela['gender'].value_counts()
```

```python
# Distribuição de data de compra
ni_data = tabela['date'].value_counts()
print(ni_data)
```

```python
# Percentual da quantidade de compras que há em uma data específica
percentual_7_11 = ((tabela['date'].value_counts() >= 7) & (tabela['date'].value_counts() <= 11)).mean() * 100
percentual_3 = (tabela['date'].value_counts() == 3).mean() * 100

print(f'mediana: {tabela['date'].value_counts().median()}')
print(f'média: {tabela['date'].value_counts().mean()}') #\n\n\n\n
print(f'Percentual da distribuição de 7 a 11: {percentual_7_11}')
print(f'percentual da distribuição igual a 3: {percentual_3}')
```

Observa-se, pela mediana, que a empresa atende, em sua maioria, aproximadamente 3 clientes por dia, tendo a uma frequência relativa (Fi) 13.8, aproximadamente, vezes maior que a frequência relativa acumulada (Fia) do intervalo discreto 7 a 11.

Portanto, vê-se uma assimetria que representa um aumento significativo de vendas em determinadas datas. Como somente os dias ditam os momentos de ápice de vendas, infere-se que haja talvez a falta de organização ou de conhecimento sobre o momento correto de ceder descontos ou semelhantes aos clientes.
